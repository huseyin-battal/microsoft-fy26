```python
from qdk import qsharp
from azure.quantum import Workspace

qsharp.init(target_profile=qsharp.TargetProfile.Base)   # donanıma göndermek için Base profile şart
program = qsharp.compile("GroverSearchAlgorithm()")

workspace = Workspace(resource_id="<Azure Quantum workspace resource ID>")
target = workspace.get_targets("rigetti.sim.qvm")

job = target.submit(program, "GroverJob", shots=100)
results = job.get_results()
```
